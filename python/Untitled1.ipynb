```python
import control as ct
import matplotlib.pyplot as plt # plots 
```

```python
L=(1.65+1.33)*1E-4
print(L)
R=0.02
print(R)
```

```python
sys = ct.tf([1],[L,R])
print(sys)
t, y = ct.step_response(sys, 0.1)
plt.plot(t, y, label='continouous-time model')
plt.legend()
plt.xlabel('time (s)');
```

```python
Ts=1E-2
dsys = ct.c2d(sys, Ts, "tustin")
print(dsys)
t, y = ct.step_response(dsys, 0.1)
plt.plot(t, y, '.-', label='discrete-time model')
plt.legend()
plt.xlabel('time (s)');
```
