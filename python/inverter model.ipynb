```python
import control as ct
from math import pi
import numpy as np
```

```python
w=2*pi*50
print(w)
```

```python
L=1.65E-3
```

```python
A = np.array([[0,w],[-w,0]])
print('A=',A);
B=np.array([[1/L,0],[0,1/L]])
print('B=',B)
C=np.array([[1,0],[0,1]])
print('C=',C)
D=0
```

```python
plant = ct.ss(A, B, C, D, outputs=['d', 'q'])
print(plant)
```

```python
Ts=1E-4
plantd = ct.c2d(plant, Ts)
print(plantd)
```
