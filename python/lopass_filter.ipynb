```python
import control as ct
import matplotlib.pyplot as plt # plots
from math import pi
```

```python
s=ct.tf('s')
a=2*pi*5
G=a/(s+a)
print(G)
```

```python
response = ct.step_response(G, T=0.2)
cplt = response.plot()
```

```python
Ts=1E-4
Gd=ct.c2d(G,Ts,'tustin')
print(Gd)
```

```python
response = ct.step_response(Gd, T=0.2)
cplt = response.plot()
```
